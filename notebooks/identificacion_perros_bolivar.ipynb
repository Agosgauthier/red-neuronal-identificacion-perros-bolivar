```python
import os
import random
from glob import glob
from itertools import combinations
import pandas as pd

random.seed(42)

base = "dataset/perros/YT-BB-Dog/train"

# Obtener perros que tengan al menos 2 imágenes
perros = []

for perro in os.listdir(base):
    ruta = os.path.join(base, perro)

    if os.path.isdir(ruta):
        imagenes = glob(os.path.join(ruta, "*.jpg"))

        if len(imagenes) >= 2:
            perros.append((perro, imagenes))

# Elegimos 20 perros para nuestro primer MVP
perros = random.sample(perros, 20)

pares = []

# -------------------------
# PARES DEL MISMO PERRO
# -------------------------
for perro, imagenes in perros:

    combinaciones = list(combinations(imagenes, 2))

    # Tomamos hasta 10 pares por perro
    muestras = random.sample(
        combinaciones,
        min(10, len(combinaciones))
    )

    for img1, img2 in muestras:
        pares.append({
            "imagen1": img1,
            "imagen2": img2,
            "label": 1
        })

# -------------------------
# PARES DE PERROS DISTINTOS
# -------------------------
for _ in range(len(pares)):

    perro1, imagenes1 = random.choice(perros)
    perro2, imagenes2 = random.choice(perros)

    while perro1 == perro2:
        perro2, imagenes2 = random.choice(perros)

    img1 = random.choice(imagenes1)
    img2 = random.choice(imagenes2)

    pares.append({
        "imagen1": img1,
        "imagen2": img2,
        "label": 0
    })

# Crear DataFrame
df_pares = pd.DataFrame(pares)

# Mezclar
df_pares = df_pares.sample(frac=1, random_state=42).reset_index(drop=True)

# Guardar
os.makedirs("dataset", exist_ok=True)

df_pares.to_csv(
    "dataset/pares_mvp.csv",
    index=False
)

print("Pares totales:", len(df_pares))
print("Mismo perro (1):", (df_pares["label"] == 1).sum())
print("Perros diferentes (0):", (df_pares["label"] == 0).sum())

print("\nPrimeros pares:")
display(df_pares.head())
```

```python
import random
import os
from glob import glob
from itertools import combinations
import pandas as pd

random.seed(42)


def crear_pares(lista_perros, pares_mismo_por_perro=10):
    pares = []

    # -------------------------
    # MISMO PERRO → label 1
    # -------------------------
    for perro, imagenes in lista_perros:

        combinaciones = list(combinations(imagenes, 2))

        cantidad = min(pares_mismo_por_perro, len(combinaciones))

        muestras = random.sample(combinaciones, cantidad)

        for img1, img2 in muestras:
            pares.append({
                "imagen1": img1,
                "imagen2": img2,
                "label": 1
            })

    # -------------------------
    # PERROS DIFERENTES → label 0
    # -------------------------
    cantidad_mismo = sum(
        1 for par in pares if par["label"] == 1
    )

    for _ in range(cantidad_mismo):

        perro1, imagenes1 = random.choice(lista_perros)
        perro2, imagenes2 = random.choice(lista_perros)

        while perro1 == perro2:
            perro2, imagenes2 = random.choice(lista_perros)

        img1 = random.choice(imagenes1)
        img2 = random.choice(imagenes2)

        pares.append({
            "imagen1": img1,
            "imagen2": img2,
            "label": 0
        })

    df = pd.DataFrame(pares)

    return df.sample(
        frac=1,
        random_state=42
    ).reset_index(drop=True)


# Crear pares de entrenamiento
df_train = crear_pares(perros_train)

# Crear pares de prueba
df_test = crear_pares(perros_test)


# Guardar
df_train.to_csv(
    "dataset/pares_train.csv",
    index=False
)

df_test.to_csv(
    "dataset/pares_test.csv",
    index=False
)


print("PARES DE ENTRENAMIENTO")
print("Total:", len(df_train))
print("Mismo perro:", (df_train["label"] == 1).sum())
print("Diferentes:", (df_train["label"] == 0).sum())

print("\nPARES DE PRUEBA")
print("Total:", len(df_test))
print("Mismo perro:", (df_test["label"] == 1).sum())
print("Diferentes:", (df_test["label"] == 0).sum())
```

```python
import cv2
import numpy as np
import pandas as pd


TAMANO = 128


def cargar_imagen(ruta):
    imagen = cv2.imread(ruta)

    if imagen is None:
        raise ValueError(f"No se pudo cargar: {ruta}")

    imagen = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    imagen = cv2.resize(imagen, (TAMANO, TAMANO))
    imagen = imagen.astype("float32") / 255.0

    return imagen


def preparar_datos(df):
    X1 = []
    X2 = []
    y = []

    for _, fila in df.iterrows():
        X1.append(cargar_imagen(fila["imagen1"]))
        X2.append(cargar_imagen(fila["imagen2"]))
        y.append(fila["label"])

    return (
        np.array(X1),
        np.array(X2),
        np.array(y, dtype="float32")
    )


X1_train, X2_train, y_train = preparar_datos(df_train)
X1_test, X2_test, y_test = preparar_datos(df_test)


print("Datos de entrenamiento:")
print("X1:", X1_train.shape)
print("X2:", X2_train.shape)
print("y :", y_train.shape)

print("\nDatos de prueba:")
print("X1:", X1_test.shape)
print("X2:", X2_test.shape)
print("y :", y_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model


def crear_red():
    entrada = layers.Input(shape=(128, 128, 3))

    x = layers.Conv2D(32, 3, activation="relu")(entrada)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(64, 3, activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(128, 3, activation="relu")(x)
    x = layers.GlobalAveragePooling2D()(x)

    x = layers.Dense(128, activation="relu")(x)

    return Model(entrada, x)


red = crear_red()

red.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model


# Dos entradas: una por cada foto
entrada1 = layers.Input(shape=(128, 128, 3))
entrada2 = layers.Input(shape=(128, 128, 3))


# Las dos imágenes pasan por la MISMA red
vector1 = red(entrada1)
vector2 = red(entrada2)


# Comparamos las características de ambas imágenes
diferencia = layers.Lambda(
    lambda x: tf.abs(x[0] - x[1])
)([vector1, vector2])


# Clasificación final
x = layers.Dense(64, activation="relu")(diferencia)
salida = layers.Dense(1, activation="sigmoid")(x)


# Crear modelo completo
modelo = Model(
    inputs=[entrada1, entrada2],
    outputs=salida
)


# Compilar
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


modelo.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2


# Red preentrenada
base_model = MobileNetV2(
    input_shape=(128, 128, 3),
    include_top=False,
    weights="imagenet"
)

# Congelamos la red preentrenada
base_model.trainable = False


# Red que extrae características
entrada = layers.Input(shape=(128, 128, 3))

# Nuestras imágenes están entre 0 y 1
# MobileNetV2 trabaja con valores entre -1 y 1
x = layers.Rescaling(
    scale=2.0,
    offset=-1.0
)(entrada)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(128, activation="relu")(x)

red_mejorada = Model(entrada, x)


# Dos imágenes
entrada1 = layers.Input(shape=(128, 128, 3))
entrada2 = layers.Input(shape=(128, 128, 3))

vector1 = red_mejorada(entrada1)
vector2 = red_mejorada(entrada2)


# Comparar características
diferencia = layers.Lambda(
    lambda x: tf.abs(x[0] - x[1])
)([vector1, vector2])


# Clasificación
x = layers.Dense(64, activation="relu")(diferencia)
salida = layers.Dense(1, activation="sigmoid")(x)


modelo_mejorado = Model(
    inputs=[entrada1, entrada2],
    outputs=salida
)


modelo_mejorado.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


modelo_mejorado.summary()
```

```python
import os
import random
from glob import glob
import pandas as pd

random.seed(42)

base = "dataset/perros/YT-BB-Dog"

train_path = os.path.join(base, "train")
test_path = os.path.join(base, "test")


def obtener_perros(ruta):
    resultado = []

    for perro in os.listdir(ruta):
        carpeta = os.path.join(ruta, perro)

        if os.path.isdir(carpeta):
            imagenes = glob(os.path.join(carpeta, "*.jpg"))

            if len(imagenes) >= 2:
                resultado.append((perro, imagenes))

    return resultado


# Obtener perros disponibles
todos_train = obtener_perros(train_path)
todos_test = obtener_perros(test_path)

# Elegir perros
perros_train_grande = random.sample(todos_train, 100)
perros_test_grande = random.sample(todos_test, 20)


def crear_pares(lista_perros, pares_mismo=20):
    pares = []

    # -------------------------
    # MISMO PERRO = 1
    # -------------------------
    for perro, imagenes in lista_perros:

        if len(imagenes) < 2:
            continue

        for _ in range(pares_mismo):
            img1, img2 = random.sample(imagenes, 2)

            pares.append({
                "imagen1": img1,
                "imagen2": img2,
                "label": 1
            })

    # -------------------------
    # PERROS DIFERENTES = 0
    # -------------------------
    cantidad_diferentes = len(pares)

    for _ in range(cantidad_diferentes):

        perro1, imagenes1 = random.choice(lista_perros)
        perro2, imagenes2 = random.choice(lista_perros)

        while perro1 == perro2:
            perro2, imagenes2 = random.choice(lista_perros)

        img1 = random.choice(imagenes1)
        img2 = random.choice(imagenes2)

        pares.append({
            "imagen1": img1,
            "imagen2": img2,
            "label": 0
        })

    df = pd.DataFrame(pares)

    return df.sample(
        frac=1,
        random_state=42
    ).reset_index(drop=True)


# Crear dataset grande
df_train_grande = crear_pares(
    perros_train_grande,
    pares_mismo=20
)

df_test_grande = crear_pares(
    perros_test_grande,
    pares_mismo=20
)


# Guardar
df_train_grande.to_csv(
    "dataset/pares_train_grande.csv",
    index=False
)

df_test_grande.to_csv(
    "dataset/pares_test_grande.csv",
    index=False
)


print("ENTRENAMIENTO")
print("Perros:", len(perros_train_grande))
print("Pares:", len(df_train_grande))
print("Mismo:", (df_train_grande["label"] == 1).sum())
print("Diferentes:", (df_train_grande["label"] == 0).sum())

print("\nPRUEBA")
print("Perros:", len(perros_test_grande))
print("Pares:", len(df_test_grande))
print("Mismo:", (df_test_grande["label"] == 1).sum())
print("Diferentes:", (df_test_grande["label"] == 0).sum())
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model


@tf.keras.utils.register_keras_serializable(package="Perros")
class AbsDifference(layers.Layer):
    def call(self, inputs):
        return tf.abs(inputs[0] - inputs[1])


# Dos entradas
entrada1_export = layers.Input(
    shape=(128, 128, 3),
    name="imagen_1"
)

entrada2_export = layers.Input(
    shape=(128, 128, 3),
    name="imagen_2"
)


# Usamos la misma red que ya entrenamos
vector1_export = red_mejorada(entrada1_export)
vector2_export = red_mejorada(entrada2_export)


# Comparación sin Lambda
diferencia_export = AbsDifference(name="diferencia")(
    [vector1_export, vector2_export]
)


# Capas finales
comparacion_export = layers.Dense(
    64,
    activation="relu",
    name="comparacion"
)(diferencia_export)

salida_export = layers.Dense(
    1,
    activation="sigmoid",
    name="salida"
)(comparacion_export)


# Modelo final exportable
modelo_exportable = Model(
    inputs=[entrada1_export, entrada2_export],
    outputs=salida_export
)


# Copiar los pesos de las capas finales del modelo que ya entrenamos
modelo_exportable.get_layer("comparacion").set_weights(
    modelo_mejorado.get_layer("dense_7").get_weights()
)

modelo_exportable.get_layer("salida").set_weights(
    modelo_mejorado.get_layer("dense_8").get_weights()
)


modelo_exportable.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


print("Modelo exportable creado correctamente.")
modelo_exportable.summary()
```

```python
# Instalar Gradio
%pip install -q gradio

import tensorflow as tf
import numpy as np
import gradio as gr


# --------------------------------------------------
# Capa utilizada por nuestro modelo
# --------------------------------------------------

@tf.keras.utils.register_keras_serializable(package="Perros")
class AbsDifference(tf.keras.layers.Layer):
    def call(self, inputs):
        return tf.abs(inputs[0] - inputs[1])


# --------------------------------------------------
# Cargar el modelo final
# --------------------------------------------------

modelo_final = tf.keras.models.load_model(
    "modelo_perros_bolivar_final.keras",
    custom_objects={"AbsDifference": AbsDifference},
    safe_mode=False
)

print("Modelo cargado correctamente.")


# --------------------------------------------------
# Preparar imagen
# --------------------------------------------------

def preparar_imagen(imagen):

    # Convertir a array de NumPy
    imagen = np.array(imagen)

    # Si tiene canal alfa, conservar solamente RGB
    if imagen.shape[-1] == 4:
        imagen = imagen[:, :, :3]

    # Redimensionar
    imagen = tf.image.resize(
        imagen,
        (128, 128)
    )

    # Convertir a valores entre 0 y 1
    imagen = tf.cast(
        imagen,
        tf.float32
    ) / 255.0

    # Agregar dimensión del batch
    imagen = tf.expand_dims(
        imagen,
        axis=0
    )

    return imagen


# --------------------------------------------------
# Comparar dos perros
# --------------------------------------------------

def comparar_perros(imagen1, imagen2):

    if imagen1 is None or imagen2 is None:
        return "⚠️ Cargá las dos imágenes."

    imagen1 = preparar_imagen(imagen1)
    imagen2 = preparar_imagen(imagen2)

    # Predicción
    prediccion = modelo_final.predict(
        [imagen1, imagen2],
        verbose=0
    )[0][0]

    probabilidad_mismo = float(prediccion)
    probabilidad_diferente = 1 - probabilidad_mismo

    if probabilidad_mismo >= 0.5:

        return (
            "🐕✅ MISMO PERRO\n\n"
            f"Probabilidad de mismo perro: "
            f"{probabilidad_mismo * 100:.2f}%"
        )

    else:

        return (
            "🐕❌ PERROS DIFERENTES\n\n"
            f"Probabilidad de perros diferentes: "
            f"{probabilidad_diferente * 100:.2f}%"
        )


# --------------------------------------------------
# Interfaz visual
# --------------------------------------------------

with gr.Blocks() as demo:

    gr.Markdown(
        """
        # 🐕 Identificación de Perros en Bolívar

        ## Comparación de dos fotografías

        Cargá dos fotografías para determinar
        si corresponden al mismo perro o a perros diferentes.
        """
    )

    with gr.Row():

        imagen1 = gr.Image(
            type="numpy",
            label="📷 Imagen 1"
        )

        imagen2 = gr.Image(
            type="numpy",
            label="📷 Imagen 2"
        )

    boton = gr.Button(
        "🔎 Comparar perros"
    )

    resultado = gr.Textbox(
        label="Resultado",
        lines=3
    )

    boton.click(
        fn=comparar_perros,
        inputs=[imagen1, imagen2],
        outputs=resultado
    )


# --------------------------------------------------
# Ejecutar interfaz
# --------------------------------------------------

demo.launch()
```
